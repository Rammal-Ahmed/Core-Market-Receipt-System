```python
# ============================================================
# CORE MARKET RECEIPT SYSTEM
# Supermarket Billing and Customer Receipt
# ============================================================

print("━" * 45)
print("CORE MARKET RECEIPT SYSTEM".center(45))
print()
print("SUPERMARKET BILLING & CUSTOMER RECEIPT".center(45))
print("━" * 45)
print(

)
# -------------------- CUSTOMER INPUT -------------------------

customer_name = input("What is your name, Sir?: ")
product_name = input("Which product would you like to purchase?: ")
product_price = float(input("Product unit price (Rs.): "))
quantity = int(input("What is the quantity, Sir?: "))
payment_method = input("Enter payment method (Cash / Card): ").lower()


# -------------------- CUSTOMER DISCOUNT ----------------------

gross_bill = quantity * product_price

if quantity >= 52:
    discount = gross_bill * 20 / 100
    customer_type = "VIP Customer"
    print("As a VIP Customer, you will receive a 20% discount")

elif quantity >= 35:
    discount = gross_bill * 10 / 100
    customer_type = "Premium Customer"
    print("As a Premium Customer, you will receive a 10% discount")

elif quantity >= 12:
    discount = gross_bill * 5 / 100
    customer_type = "Regular Customer"
    print("As a Regular Customer, you will receive a 5% discount")

else:
    discount = 0
    customer_type = "Regular Customer"


discounted_price = gross_bill - discount


# -------------------- PAYMENT DISCOUNT ----------------------

payment_discount = 0

if (
    (gross_bill >= 100000 and customer_type == "VIP Customer")
    or gross_bill >= 200000
):

    if payment_method == "card":
        payment_discount = gross_bill * 25 / 100
        print("You received a 25% card discount")

    else:
        payment_discount = 0
        print("No card discount applied to this purchase")

else:
    print("You are not eligible for the additional payment discount")


final_price = gross_bill - discount - payment_discount


# -------------------- SHIPPING CHARGES ----------------------

if final_price >= 100000:
    shipping = 0
    print("Congratulations! You received free shipping")

elif final_price >= 50000:
    shipping = 250
    print("You received a 50% shipping discount")

else:
    shipping = 500
    print("Shipping charges: Rs. 500 only")


total_bill = final_price + shipping


# -------------------- CUSTOMER RECEIPT ----------------------

print()
print("=" * 40)
print("CORE MARKET RECEIPT".center(40))
print("CUSTOMER SALES RECEIPT".center(40))
print("=" * 40)
print()

print("Customer Name:          ", customer_name)
print()
print("Product Description:    ", product_name)
print()
print("Unit Price (Rs.):       ", product_price)
print()
print("Quantity Purchased:     ", quantity)
print()
print("Gross Amount (Rs.):     ", gross_bill)
print()
print("Purchase Discount (Rs.):", discount)
print()
print("Payment Discount (Rs.): ", payment_discount)
print()
print("Net Amount (Rs.):       ", final_price)
print()
print("Shipping Charges (Rs.): ", shipping)
print()
print("Total Payable (Rs.):    ", total_bill)

print()
print("-" * 40)
print("TRANSACTION COMPLETED!".center(40))
print("THANK YOU FOR SHOPPING".center(40))
print("SEE YOU IN THE NEXT REPOSITORIES..!".center(40))
print("=" * 40)
```
